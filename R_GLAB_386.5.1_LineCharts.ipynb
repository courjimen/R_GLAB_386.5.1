# R_GLAB_386.5.1 Line Charts

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

#create a line plot with a dashed dot line style
plt.plot(x, y, linestyle='-.', marker='o')
plt.show()
```

```python
#object oriented approach
fig, ax = plt.subplots()

ax.plot(x,y, linestyle='-.', marker='*', markersize='15')
plt.show()
```

```python
#plot the dotted line style
plt.plot(x, y, linestyle=':')
```

```python
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
#create a dashed line with diamond markers
plt.plot(y, linestyle='--', marker='D') # when you only plot y variable it uses the indexes for the x
plt.show()
```

```python
#using the opposite approach create dashdot line and square markers
fig, ax = plt.subplots()

ax.plot(y, linestyle='-.', marker='s')
plt.show()
```

```python
#use any approach, diamond marker colored red, marker width
plt.plot(y, color='black', marker='D', mfc='pink', ms='10', mew='2', mec='purple')
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
#create a lineplot, custom color, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=':', color='purple', marker='D', mfc='black', mew='2')
plt.show()
```
