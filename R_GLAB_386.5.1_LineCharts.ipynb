# R_GLAB_386.5.1 Line Charts

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

#create a line plot with a dashed dot line style
plt.plot(x, y, linestyle='-.', marker='o')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x,y, linestyle='-.', marker='*', markersize='15')
plt.show()
```
