### Taking closer look at education and final weight
- goal: to observe the amount of population that is observed in each education level

```python
# grab values from the columns we want
df['education'].value_counts()
```

```python
subset = df.groupby('education')['fnlwgt'].sum().reset_index()
subset
```

```python
# plot our information from education
plt.plot(
    subset['education'],
    subset['fnlwgt']
)
plt.title('Amount of Educated Population')
plt.xlabel('Education')
plt.ylabel('Population (Billions)')
plt.xticks(rotation=90)
plt.grid(alpha=0.3)
plt.show()
```

```python
#subset of median age per education group
subset = df.groupby('education')['age'].median().reset_index()
subset
```

```python
plt.plot(
    subset['education'],
    subset['age'],
    marker='o'
)
plt.title('Median Age and Education Level')
plt.xticks(rotation=90)
plt.xlabel('Highest Education')
plt.ylabel('Median Age')
```
